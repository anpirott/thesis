## Final conclusion on the filters

Filtering on the log_g => not useful as the results of the two parameters go in opposite directions (when one improves, the other worsens)

Filtering on the age => age_filtered_bigger_7 improves the results of both parameters compared to Base and might be useful to use if we are fine with losing the possibility to predict for stars below 10.000.000 years

Filtering on the mass => not useful to use this filtering as it gives similar results as Base while being restricted in the kind of stars it can predict
