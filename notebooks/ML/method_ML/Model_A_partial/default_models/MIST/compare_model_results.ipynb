#### Conclusions on mass filtering

mass_smaller_3 is always worse than Base.

If reducing worst-case errors is the priority, mass_smaller_14 is preferable; if average accuracy matters more, Base is better.

mass_smaller_30 is a middle ground between Base and mass_smaller_14 but it does not offer enough improvements to be interesting to use.

#### Best models for the mass parameter

The three best performing models in order for the mass parameter are Random forest, Decision tree and KNN.
