### Mass parameter


```python
compare_metrics(path, 
                output_parameters=["mass"],
                physical_models=["PARSEC"],
                data_filters=["Base", "mass_smaller_3"])
```

Every comparison is proportional to the mass.

Linear regression : 
- mass_smaller_3 : better results than Base on every metric except MAX_ER

Decision tree : 
- mass_smaller_3 : worse results than Base on every metric.

KNN : 
- mass_smaller_3 : worse results than Base on every metric except MAX_ER.

Random forest : 
- mass_smaller_3 : worse results than Base on every.

XGBoost : 
- mass_smaller_3 : worse results than Base on every metric except MAX_ER.

MLP : 
- mass_smaller_3 : worse results than Base on every metric except MAX_ER.

#### Conclusions on mass

Filtering with mass_smaller_3 gives worse results on every metric except for MAX_ER and the results of linear regression.


#### Best models for the mass parameter

The three best performing models in order for the mass parameter are Random forest, Decision tree and XGBoost.

&emsp; => important to note that the MedAE for random forest and decision tree is very close to 0, actually 0 for decision tree. \
&emsp; &emsp; This could be due to the fact that the PARSEC grids are discrete and the tree models accomodate well with this structure. \
&emsp; &emsp; These models might not be able to generalise enough for real-world data if they follow the grid too closely. \
