## Import

```python
import joblib
```

```python
pre_path = "../../../../../../"
physical_model = "MIST"
path_to_final_models_B = 'model/final_models/Model_B/MIST/'
path_to_final_models_B_scaler = 'model/final_models/Model_B/PARSEC/scalers/quantile_scaler.pkl'
```

```python
# star_mass en masse solaire, log_age en log_10 de l'age en année, metallicity en dex, 
# log_T_eff en log_10 de la température en Kelvin, log_g en log_10 de la gravité en cm/s^2, 
# log_R en log_10 du rayon en rayon solaire, log_L en log_10 de la luminosité en luminosité solaire

sun_params = {"star_mass" : 1, "log_age" : 9.65, "metallicity" : 0, "log_Teff" : 3.76, "log_g" : 4.44, "log_R" : 0, "log_L" : 0}

arcturus_params = {"star_mass" : 1.08, "log_age" : 9.89, "metallicity" : -0.5, "log_Teff" : 3.63, "log_g" : 1.66, "log_R" : 1.39, "log_L" : 2.23}
```

```python
def test_second_model(log_age, metallicity, mass, model_B_path, scaler_B_path=None):
    model_B = joblib.load(model_B_path)

    model_B_inputs = [[log_age, metallicity, mass]]
    if scaler_B_path is not None: # apply a scaling to the data if provided
        model_B_scaler = joblib.load(scaler_B_path)
        model_B_inputs = model_B_scaler.transform(model_B_inputs)
    # Predict the effective temperature, surface gravity and radius of the secondary star
    log_Teff2, log_g2, log_R2 = model_B.predict(model_B_inputs).flatten()

    return log_Teff2, log_g2, log_R2
```
