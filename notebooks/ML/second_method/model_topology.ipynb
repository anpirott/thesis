```python
%matplotlib ipympl
# en X la t_eff
# en y la log_g
# en z l'erreur entre le m2 et le m2 voulu

# je prends 2 valeurs et je les mets dans le modèle, je calcule l'erreur avec le m2 voulu et je continue avec 2 autre valeurs

# faire aussi les plots en 2d avec un des deux paramètres fixé

age = 7.0
metallicity = 0.0
single_isochrone = phase_filtered_iso_df.loc[((phase_filtered_iso_df['metallicity'] == metallicity) & (phase_filtered_iso_df['log10_isochrone_age_yr'] == age))].reset_index()

num_values = 39 # how many values we want for the graph - 1
precision = 10000

min_t_eff_values, max_t_eff_values = int(min(single_isochrone.log_Teff)*precision), int(max(single_isochrone.log_Teff)*precision)
step_t_eff_values = int((max(single_isochrone.log_Teff) - min(single_isochrone.log_Teff))*precision/num_values)
t_eff_values = [i/precision for i in range(min_t_eff_values, max_t_eff_values, step_t_eff_values)]
# print(min(single_isochrone.log_Teff), max(single_isochrone.log_Teff))
# print(t_eff_values)
# print(len(t_eff_values))

min_log_g_values, max_log_g_values = int(min(single_isochrone.log_g)*precision), int(max(single_isochrone.log_g)*precision)
step_log_g_values = int((max(single_isochrone.log_g) - min(single_isochrone.log_g))*precision/num_values)
log_g_values = [i/precision for i in range(min_log_g_values, max_log_g_values, step_log_g_values)]
# print(min(single_isochrone.log_g), max(single_isochrone.log_g))
# print(log_g_values)
# print(len(log_g_values))

ml_model = joblib.load(r"C:\Users\antoi\Code\unif\MA2\thesis\model\fully_trained\primary_XGB.pkl")
# star_mass1, log_R1 = ml_model.predict([[age, metallicity, XXX, XXX]]).flatten()
print(min(single_isochrone.star_mass), max(single_isochrone.star_mass))
desired_m2 = 5
err = []
abs_err = []
for t_eff in t_eff_values:
    temp_err = []
    temp_abs_err = []
    for log_g in log_g_values:
        star_mass2, log_R2 = ml_model.predict([[age, metallicity, t_eff, log_g]]).flatten()
        # print(star_mass2) # masse négative??? avec age = 5.5, métallicité = 0.0 et desired_m2 = 100
        # sys.exit(0) # TODO! checkez si c'est tous les modèles ou juste celui que j'utilise qui n'est pas complètement fine-tuned
        temp_err.append(star_mass2 - desired_m2) # TODO! le modèle donne exactement les mêmes valeurs pour des t_eff et log_g différents (même chose qu'au dessus)
        temp_abs_err.append(abs(star_mass2 - desired_m2))
    err.append(temp_err)
    abs_err.append(temp_abs_err)
err = np.array(err)
abs_err = np.array(abs_err)

# fig = plt.figure(figsize=plt.figaspect(0.5))
fig = plt.figure(figsize=(10, 10))
X, Y = np.meshgrid(t_eff_values, log_g_values)

# error
ax = fig.add_subplot(2, 2, 1, projection='3d')
ax.plot_wireframe(X, Y, err, rstride=5, cstride=5)

ax.set_xlabel('$T_{eff}$')
ax.set_ylabel('$log(g)$')
ax.set_zlabel('$Error$')

ax = fig.add_subplot(2, 2, 2, projection='3d')
surf = ax.plot_surface(X, Y, err, rstride=1, cstride=1, cmap=cm.coolwarm,
                       linewidth=0, antialiased=False)
fig.colorbar(surf, shrink=0.5, aspect=10)

ax.set_xlabel('$T_{eff}$')
ax.set_ylabel('$log(g)$')
ax.set_zlabel('$Error$')

# absolute error
ax = fig.add_subplot(2, 2, 3, projection='3d')
ax.plot_wireframe(X, Y, abs_err, rstride=5, cstride=5)

ax.set_xlabel('$T_{eff}$')
ax.set_ylabel('$log(g)$')
ax.set_zlabel('$Absolute error$')

ax = fig.add_subplot(2, 2, 4, projection='3d')
surf = ax.plot_surface(X, Y, abs_err, rstride=1, cstride=1, cmap=cm.coolwarm,
                       linewidth=0, antialiased=False)
fig.colorbar(surf, shrink=0.5, aspect=10)

ax.set_xlabel('$T_{eff}$')
ax.set_ylabel('$log(g)$')
ax.set_zlabel('$Absolute error$')

plt.show()
```
