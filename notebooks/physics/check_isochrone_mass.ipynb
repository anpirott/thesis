```python
iso_handler = Iso_data_handler(path_to_data, 
                              ['log10_isochrone_age_yr', 'log_Teff', 'log_g', 'phase', 'metallicity', 'star_mass', 'log_R', 'log_L'], 
                              physical_model, reclassify=True)

iso_df = iso_handler.get_isochrone_dataframe()
```
